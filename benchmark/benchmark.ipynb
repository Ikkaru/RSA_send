# Benchmark Komparatif Keadilan (Apple-to-Apple) Murni Python

**Judul:** *Implementasi Algoritma RSA-KEM Berbasis Optimasi CRT pada Transport Layer Security (TLS) untuk Pertukaran Kunci AES yang Aman*

### Deskripsi
Notebook ini menggunakan **Pure Python Implementation** untuk semua skema (RSA-KEM, RSA-OAEP, PKCS #1 v1.5). Semua fungsi eksponensiasi modular ($C^d \bmod N$ dan optimasi CRT) dijalankan menggunakan fungsi standar `pow()` di Python, sementara seluruh struktur *padding*, *unpadding*, dan fungsi *hash* (MGF1) diprogram secara manual.

Tujuannya adalah mengeliminasi bias *C-backend* OpenSSL, sehingga kita dapat membandingkan **kompleksitas algoritma murni** (terutama *overhead padding* vs *Key Derivation Function*) secara adil.

```python
import time
import os
import hashlib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from cryptography.hazmat.primitives.asymmetric import rsa
from cryptography.hazmat.primitives.kdf.hkdf import HKDF
from cryptography.hazmat.primitives import hashes

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.dpi'] = 300
print('Library berhasil diimpor.')
```

```python
def xor_bytes(a, b):
    return bytes(x ^ y for x, y in zip(a, b))

def mgf1(seed, length):
    '''Mask Generation Function (MGF1) berbasis SHA-256'''
    counter = 0
    output = b''
    while len(output) < length:
        c = counter.to_bytes(4, byteorder='big')
        output += hashlib.sha256(seed + c).digest()
        counter += 1
    return output[:length]

class PurePythonRSAEngine:
    def __init__(self, key_size=2048):
        self.key_size = key_size
        self.private_key = rsa.generate_private_key(public_exponent=65537, key_size=key_size)
        pn = self.private_key.private_numbers()
        self.n, self.e, self.d = pn.public_numbers.n, pn.public_numbers.e, pn.d
        self.p, self.q = pn.p, pn.q
        self.dp, self.dq, self.qinv = pn.dmp1, pn.dmq1, pn.iqmp

    # --- CORE MATH ---
    def rsa_ep(self, m_int):
        return pow(m_int, self.e, self.n)

    def rsa_dp_non_crt(self, c_int):
        return pow(c_int, self.d, self.n)

    def rsa_dp_crt(self, c_int):
        m1 = pow(c_int, self.dp, self.p)
        m2 = pow(c_int, self.dq, self.q)
        h = (self.qinv * (m1 - m2)) % self.p
        return m2 + h * self.q

    # --- 1. RSA-KEM ---
    def kem_encapsulate(self):
        byte_len = self.key_size // 8
        x_int = int.from_bytes(os.urandom(byte_len - 1), 'big') % self.n
        c_int = self.rsa_ep(x_int)
        hkdf = HKDF(algorithm=hashes.SHA256(), length=32, salt=None, info=b'TLS')
        k = hkdf.derive(x_int.to_bytes(byte_len, 'big'))
        return c_int, k

    def kem_decapsulate(self, c_int, use_crt=True):
        byte_len = self.key_size // 8
        x_int = self.rsa_dp_crt(c_int) if use_crt else self.rsa_dp_non_crt(c_int)
        hkdf = HKDF(algorithm=hashes.SHA256(), length=32, salt=None, info=b'TLS')
        return hkdf.derive(x_int.to_bytes(byte_len, 'big'))

    # --- 2. RSA-OAEP (SHA-256) ---
    def oaep_encrypt(self, m):
        k = self.key_size // 8
        hlen = 32
        lhash = hashlib.sha256(b'').digest()
        ps = b'\x00' * (k - len(m) - 2 * hlen - 2)
        db = lhash + ps + b'\x01' + m
        seed = os.urandom(hlen)
        db_mask = mgf1(seed, k - hlen - 1)
        masked_db = xor_bytes(db, db_mask)
        seed_mask = mgf1(masked_db, hlen)
        masked_seed = xor_bytes(seed, seed_mask)
        em = b'\x00' + masked_seed + masked_db
        return self.rsa_ep(int.from_bytes(em, 'big'))

    def oaep_decrypt(self, c_int, use_crt=True):
        k = self.key_size // 8
        hlen = 32
        em_int = self.rsa_dp_crt(c_int) if use_crt else self.rsa_dp_non_crt(c_int)
        em = em_int.to_bytes(k, 'big')
        _, masked_seed, masked_db = em[0:1], em[1:1+hlen], em[1+hlen:]
        seed_mask = mgf1(masked_db, hlen)
        seed = xor_bytes(masked_seed, seed_mask)
        db_mask = mgf1(seed, k - hlen - 1)
        db = xor_bytes(masked_db, db_mask)
        idx = db.find(b'\x01', hlen)
        return db[idx+1:]

    # --- 3. PKCS #1 v1.5 ---
    def pkcs_encrypt(self, m):
        k = self.key_size // 8
        ps_len = k - len(m) - 3
        ps = bytearray()
        while len(ps) < ps_len:
            b = os.urandom(1)
            if b != b'\x00': ps.extend(b)
        em = b'\x00\x02' + bytes(ps) + b'\x00' + m
        return self.rsa_ep(int.from_bytes(em, 'big'))

    def pkcs_decrypt(self, c_int, use_crt=True):
        k = self.key_size // 8
        em_int = self.rsa_dp_crt(c_int) if use_crt else self.rsa_dp_non_crt(c_int)
        em = em_int.to_bytes(k, 'big')
        idx = em.find(b'\x00', 2)
        return em[idx+1:]
```

```python
engine = PurePythonRSAEngine(key_size=2048)
aes_key = os.urandom(32)
iterations = 300
results = {'Scheme': [], 'Role': [], 'Operation': [], 'Time_ms': []}

print('Menjalankan pengujian (apple-to-apple) di interpreter Python...')

# 1. KEM (CRT)
for _ in range(iterations):
    t0 = time.perf_counter()
    c_int, _ = engine.kem_encapsulate()
    results['Scheme'].append('RSA-KEM (CRT)')
    results['Role'].append('Client')
    results['Operation'].append('Encapsulation')
    results['Time_ms'].append((time.perf_counter() - t0) * 1000)
    
    t0 = time.perf_counter()
    engine.kem_decapsulate(c_int, use_crt=True)
    results['Scheme'].append('RSA-KEM (CRT)')
    results['Role'].append('Server')
    results['Operation'].append('Decapsulation')
    results['Time_ms'].append((time.perf_counter() - t0) * 1000)

# 2. KEM (Non-CRT)
for _ in range(iterations):
    c_int, _ = engine.kem_encapsulate()
    t0 = time.perf_counter()
    engine.kem_decapsulate(c_int, use_crt=False)
    results['Scheme'].append('RSA-KEM (Non-CRT)')
    results['Role'].append('Server')
    results['Operation'].append('Decapsulation')
    results['Time_ms'].append((time.perf_counter() - t0) * 1000)

# 3. OAEP (Server menggunakan CRT agar adil dengan skenario riil)
for _ in range(iterations):
    t0 = time.perf_counter()
    c_int = engine.oaep_encrypt(aes_key)
    results['Scheme'].append('RSA-OAEP (CRT)')
    results['Role'].append('Client')
    results['Operation'].append('Encryption')
    results['Time_ms'].append((time.perf_counter() - t0) * 1000)
    
    t0 = time.perf_counter()
    engine.oaep_decrypt(c_int, use_crt=True)
    results['Scheme'].append('RSA-OAEP (CRT)')
    results['Role'].append('Server')
    results['Operation'].append('Decryption')
    results['Time_ms'].append((time.perf_counter() - t0) * 1000)

# 4. PKCS #1 v1.5 (Server menggunakan CRT agar adil)
for _ in range(iterations):
    t0 = time.perf_counter()
    c_int = engine.pkcs_encrypt(aes_key)
    results['Scheme'].append('PKCS #1 v1.5 (CRT)')
    results['Role'].append('Client')
    results['Operation'].append('Encryption')
    results['Time_ms'].append((time.perf_counter() - t0) * 1000)
    
    t0 = time.perf_counter()
    engine.pkcs_decrypt(c_int, use_crt=True)
    results['Scheme'].append('PKCS #1 v1.5 (CRT)')
    results['Role'].append('Server')
    results['Operation'].append('Decryption')
    results['Time_ms'].append((time.perf_counter() - t0) * 1000)

df_raw = pd.DataFrame(results)
print('Selesai!')
```

```python
summary_df = df_raw.groupby(['Scheme', 'Role', 'Operation'])['Time_ms'].agg(['mean', 'std', 'min', 'max']).reset_index()
summary_df['Ops_Per_Sec'] = 1000.0 / summary_df['mean']
summary_df.columns = ['Skema', 'Sisi (Role)', 'Operasi', 'Rata-rata (ms)', 'Std Dev (ms)', 'Min (ms)', 'Max (ms)', 'Throughput (ops/s)']
display(summary_df.round(4))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

server_data = df_raw[df_raw['Role'] == 'Server']
server_summary = summary_df[summary_df['Sisi (Role)'] == 'Server']

sns.barplot(data=server_data, x='Scheme', y='Time_ms', ax=axes[0], palette='Blues_r', capsize=0.1)
axes[0].set_title('Execution Time: Encapsulation Decapsulation', fontsize=12, fontweight='bold', pad=15)
axes[0].set_ylabel('Waktu Eksekusi (ms)', fontsize=11)
axes[0].set_xlabel('Skema Pertukaran Kunci', fontsize=11)
axes[0].tick_params(axis='x', rotation=15)

for p in axes[0].patches:
    height = p.get_height()
    if height > 0:
        axes[0].annotate(f'{height:.2f} ms', (p.get_x() + p.get_width() / 2., height),
                         ha='center', va='bottom', xytext=(0, 4), textcoords='offset points', 
                         fontsize=10, fontweight='bold')

sns.barplot(data=server_summary, x='Skema', y='Throughput (ops/s)', ax=axes[1], palette='Greens_r')
axes[1].set_title('Server Throughput (Operations / Second)', fontsize=12, fontweight='bold', pad=15)
axes[1].set_ylabel('Kapasitas Handshake (ops/s)', fontsize=11)
axes[1].set_xlabel('Skema Pertukaran Kunci', fontsize=11)
axes[1].tick_params(axis='x', rotation=15)

for p in axes[1].patches:
    height = p.get_height()
    if height > 0:
        axes[1].annotate(f'{height:.1f} ops/s', (p.get_x() + p.get_width() / 2., height),
                         ha='center', va='bottom', xytext=(0, 4), textcoords='offset points', 
                         fontsize=10, fontweight='bold')

plt.subplots_adjust(wspace=0.35, bottom=0.2)
plt.savefig('rsa_kem_fair_benchmark.png', dpi=300, bbox_inches='tight')
plt.show()
```
